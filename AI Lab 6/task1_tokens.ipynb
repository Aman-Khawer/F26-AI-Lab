```python

import re
from sentence_transformers import SentenceTransformer

# Different text examples
texts = [
    "Hello, how are you?",
    "I can't login.",
    "My email is student@gmail.com",
    "My course is CS4944",
    "",
    "یہ ایک مثال ہے"
]

# Compare two tokenization methods
for text in texts:

    print("\nOriginal Text:", text)

    # Method 1: Whitespace splitting
    words1 = text.split()

    print("Whitespace Tokens:", words1)
    print("Whitespace Count:", len(words1))

    # Method 2: Regex tokenization
    words2 = re.findall(r"\b\w+\b", text)

    print("Regex Tokens:", words2)
    print("Regex Count:", len(words2))


# Load pretrained model
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# Model tokenizer on two English examples
print("\nModel Tokenizer")

for text in texts[:2]:

    tokens = model.tokenizer.tokenize(text)

    print("\nOriginal Text:", text)
    print("Model Tokens:", tokens)
    print("Model Token Count:", len(tokens))
```
